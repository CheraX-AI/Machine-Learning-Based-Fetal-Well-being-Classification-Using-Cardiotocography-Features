```python
subtypes = ["F_ED_N", "F_VD_N", "F_LD_N", "F_PD_N"]

# Only compare rows with all five values available
columns = ["F_D_N"] + subtypes
data = df[columns].dropna().copy()

data["calculated_total"] = data[subtypes].sum(axis=1)

matches = data["F_D_N"] == data["calculated_total"]

print("Rows checked:", len(data))
print("Matching rows:", matches.sum())
print("Non-matching rows:", (~matches).sum())
print("Match percentage:", round(matches.mean() * 100, 2), "%")
```
